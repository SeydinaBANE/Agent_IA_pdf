```python
from transformers import pipeline

qa = pipeline(
    "question-answering",
    model="etalab-ia/camembert-base-squadFR-fquad-piaf",
    tokenizer="etalab-ia/camembert-base-squadFR-fquad-piaf",
    use_fast=False  # 🔧 important pour éviter les erreurs de conversion
)

context = "La Tour Eiffel est une tour de fer située à Paris. Elle a été construite en 1889."
question = "Où se trouve la Tour Eiffel ?"

result = qa(question=question, context=context)
print("Réponse :", result["answer"])
```
